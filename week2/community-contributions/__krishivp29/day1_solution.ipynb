```python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display, update_display
```

```python
load_dotenv(override = True)

openai_api_key = os.getenv('OPENAI_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")

openai = OpenAI()
```

```python
# 3 way conversation between LLMs

luna_system = """
You are Ms.Luna, a person who is very arguementataive; you disagree with anything in the conversation and challenge everything, in a snarky way.
You are in a conversation with Mr.OSS and Mr.X.
"""

oss_system = """
You are Mr.OSS, the mayor and you try to agree with everything the other people say or find common ground, if they disagree, like a politician.
You are in a conversation with Mr.X and Ms.Luna.
"""

x_system = """
You are Mr.X, an anyonymous spy and your main goal is to get Ms.Luna and Mr.OSS to be friends to save your country.
You are in a conversation with Ms.Luna and Mr.X
"""
```

```python
def call_luna(chat_history):
    luna_user = f"""
    You are Ms.Luna, in a conversation with Mr.OSS and Mr.X.
    The conversation so far is as follows:
    {chat_history}

    Now with this, respond with what you would like to say next, as Ms.Luna starting with 'Ms:Luna: '
    """

    messages = [
        {"role": "system", "content": luna_system},
        {"role": "user", "content": luna_user}
        ]
    response = openai.chat.completions.create(model="gpt-6-luna", reasoning_effort="none", messages=messages)
    return response.choices[0].message.content

def call_oss(chat_history):
    oss_user = f"""
    You are Mr.OSS, in a conversation with Mr.X and Ms.Luna.
    The conversation so far is as follows:
    {chat_history}

    Now with this, respond with what you would like to say next, as Mr.OSS starting with 'Mr:Oss: '
    """

    messages = [
        {"role": "system", "content": oss_system},
        {"role": "user", "content":oss_user}
        ]
    response = openai.chat.completions.create(model="gpt-6-luna", reasoning_effort="none", messages=messages)
    return response.choices[0].message.content

def call_x(chat_history):
    x_user = f"""
    You are Ms.X, in a conversation with Ms.Luna and Mr.OSS.
    The conversation so far is as follows:
    {chat_history}

    Now with this, respond with what you would like to say next, as Mr.X starting with 'Mr.X: '
    """

    messages = [
        {"role": "system", "content": x_system},
        {"role": "user", "content": x_user}
        ]
    response = openai.chat.completions.create(model="gpt-6-luna", reasoning_effort="none", messages=messages)
    return response.choices[0].message.content  
```

```python
def run_convo():
    chat_history = "Mr.Oss: Hi there"

    dp = display(Markdown(chat_history), display_id = True)
    for i in range(5):
        if i != 0:
            oss_response = call_oss(chat_history)
            chat_history += f"\n\n{oss_response}"
            update_display(Markdown(chat_history), display_id = dp.display_id)
        
        luna_response = call_luna(chat_history)
        chat_history += f"\n\n{luna_response}"
        update_display(Markdown(chat_history), display_id = dp.display_id)

        x_response = call_x(chat_history)
        chat_history += f"\n\n{x_response}"
        update_display(Markdown(chat_history), display_id = dp.display_id)
```

```python
run_convo()
```
